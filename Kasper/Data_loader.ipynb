```python
import pandas as pd

# --- Load CSVs ---
patients = pd.read_csv("../TYKS_data/patients.csv")
cultures = pd.read_csv("../TYKS_data/cultures.csv")
isolates = pd.read_csv("../TYKS_data/isolates.csv")
suscept = pd.read_csv("../TYKS_data/susceptibility.csv")
antibiotics = pd.read_csv("../TYKS_data/antibiotics.csv")
wards = pd.read_csv("../TYKS_data/wards.csv")

# --- Merge datasets step-by-step ---

# 1. Add patient info to cultures
df = cultures.merge(
    patients,
    on="patient_id",
    how="left"
)

# 2. Add isolate info to cultures
df = df.merge(
    isolates,
    on="culture_id",
    how="left",
    suffixes=("", "_iso")
)

# 3. Add susceptibility results to isolates
df = df.merge(
    suscept,
    on=["culture_id", "isolate_id"],
    how="left",
    suffixes=("", "_ab")
)

# 4. Add antibiotic metadata
df = df.merge(
    antibiotics,
    left_on="antibiotic",
    right_on="antibiotic",
    how="left"
)

# 5. Add ward metadata
df = df.merge(
    wards,
    left_on="ward_code",
    right_on="antibiotic",   # <-- FIX: wards.csv is not keyed by ward_code
    how="left"
)

# NOTE:
# Your wards.csv contains antibiotic metadata, not ward metadata.
# The correct ward metadata is inside patients.csv documentation,
# so you likely want to merge cultures.ward_code with a separate wards file.
# If you have a real wards.csv, replace the merge above with:
#
# df = df.merge(wards, on="ward_code", how="left")

# --- Final dataframe ---
print(df.head())
print(df.shape)
```
